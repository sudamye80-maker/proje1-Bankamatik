```python
class Bankahesabı:
    def __init__(self, hesap_no, hesap_sahibi, baslangic_bakiye):
        self.hesap_no = hesap_no
        self.hesap_sahibi = hesap_sahibi
        self.bakiye = baslangic_bakiye
        self.islem_gecmisi = []
        print(f"Hesap Açıldı: {self.hesap_sahibi}, Hesap No: {self.hesap_no}")
        if baslangic_bakiye > 0:
            self.islem_gecmisi.append(f"Hesap açıldı. Başlangıç bakiyesi: {baslangic_bakiye} TL.")

    def bakiye_sorgulama(self):
        print(f"{self.hesap_sahibi} adlı kişinin bakiyesi: {self.bakiye} TL.")
        return self.bakiye

    def para_yatirma(self, miktar):
        if miktar > 0:
            self.bakiye += miktar
            self.islem_gecmisi.append(f"{miktar} TL para yatırıldı.")
            print(f"{miktar} TL yatırıldı. Güncel bakiye: {self.bakiye} TL.")
        else:
            print(f"Geçersiz miktar: {miktar}. Lütfen pozitif bir değer girin.")

    def para_cek(self, miktar):
        if miktar > 0:
            if self.bakiye >= miktar:
                self.bakiye -= miktar
                self.islem_gecmisi.append(f"{miktar} TL para çekildi.")
                print(f"{miktar} TL para çekildi. Güncel bakiye: {self.bakiye} TL.")
            else:
                print(f"Bakiyeniz Yetersiz!!! Çekmek istediğiniz miktar: {miktar} TL, Mevcut bakiye: {self.bakiye} TL.")
        else:
            print(f"Geçersiz miktar: {miktar}. Lütfen pozitif bir değer girin.")

    def havale_gonder(self, alici_hesap, miktar):
        if miktar > 0:
            if self.bakiye >= miktar:
                self.bakiye -= miktar
                alici_hesap.bakiye += miktar
                self.islem_gecmisi.append(f"{alici_hesap.hesap_sahibi} adlı kişiye {miktar} TL havale gönderildi.")
                alici_hesap.islem_gecmisi.append(f"{self.hesap_sahibi} adlı kişiden {miktar} TL havale alındı.")
                print(f"Havale gönderildi: {miktar} TL. Güncel bakiye: {self.bakiye} TL.")
                print(f"Gönderilen hesap: {self.hesap_sahibi}, Alıcı hesap: {alici_hesap.hesap_sahibi}.")
            else:
                print(f"Yetersiz Bakiye: {self.bakiye} TL. Havale gönderilemedi.")
        else:
            print("Geçersiz miktar!!! Lütfen pozitif bir değer girin.")

    def islem_gecmisi_goster(self):
        print(f"\n--- {self.hesap_sahibi} - İşlem Geçmişi ---")
        print(f"Hesap No: {self.hesap_no}")
        print(f"Bakiye: {self.bakiye} TL")
        print(f"İşlem sayısı: {len(self.islem_gecmisi)}")
        print("İSLEMLER:")
        for i, islem in enumerate(self.islem_gecmisi):
            print(f"{i+1}. işlem: {islem}")


# Test code outside the class
hesap1 = Bankahesabı("TR001", "Ahmet Yılmaz", 1000)
hesap2 = Bankahesabı("TR002", "Ayşe Demir", 500)
hesap3 = Bankahesabı("TR003", "Mehmet Can", 2000)

hesap3.bakiye_sorgulama()
hesap1.para_yatirma(2000)
hesap2.para_yatirma(1500)
hesap1.para_cek(500)
hesap1.havale_gonder(hesap2, 1500)

hesap1.islem_gecmisi_goster()
hesap2.islem_gecmisi_goster()
hesap3.islem_gecmisi_goster()
```
